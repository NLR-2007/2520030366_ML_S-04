Bitwise-Operators--> Works on Binary values.

```python
# Bitwise Operators in Python

a = 5      # Binary: 0101
b = 3      # Binary: 0011

print("a =", a)
print("b =", b)

# 1. Bitwise AND (&)
print("\n1. AND (&)")
print("a & b =", a & b)     # 0101 & 0011 = 0001 = 1

# 2. Bitwise OR (|)
print("\n2. OR (|)")
print("a | b =", a | b)     # 0101 | 0011 = 0111 = 7

# 3. Bitwise XOR (^)
print("\n3. XOR (^)")
print("a ^ b =", a ^ b)     # 0101 ^ 0011 = 0110 = 6

# 4. Bitwise NOT (~)
print("\n4. NOT (~)")
print("~a =", ~a)

# 5. Left Shift (<<)
print("\n5. Left Shift (<<)")
print("a << 1 =", a << 1)   # 0101 -> 1010 = 10

# 6. Right Shift (>>)
print("\n6. Right Shift (>>)")
print("a >> 1 =", a >> 1)   # 0101 -> 0010 = 2
```

```python
import random
jackpot = random.randint(1,100)
print("NLR--Number Guessing Game(NGG)")
guess = int(input("Guess the number: "))
count = 1;
while guess!=jackpot:
  if guess<jackpot:
    print("Guess Higher")
  else:
    print("Guess Lower")
  guess = int(input("Guess the number: "))
  count +=1
print("Correct. You gussed right one.")
print("You gussed in",count,"no of steps.")
```

<b>Aliasing: </b> Giving another name to same object in the memory

```python
a = 5
b = a  #same memory location A and B points.
id(b)
```

```python
del a
b
```

```python
#The point is the if a = 5, and a = b and a = 6, then what is the
#value of b.
a = 5  #--> 1st value
b = a  # a value is assigned to b
a = 6  # a value is changed to 6 but still b will be refered 5.
print(b)  # it will not point to 6 as earlier we made it point to 5.
```

<b>Garbage Collection: </b> Removing of objects from memory which no longer begin used.

```python
a = 10
b = a
c = b  #now here we have did aliasing

del a
del b
del c  #the varable a , b, c are deleted. But the value "10"--> was not deleted
#So here GC checks the values and if none of var is pointing to them values it
#delete them. So this process is called Garbage Collector.
```
